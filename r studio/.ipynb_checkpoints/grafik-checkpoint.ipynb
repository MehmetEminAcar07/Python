```r
plot(satışlar$takım, type="o", col="blue", xlab="aylar", ylab="satışlar", main="aylık satış miktarları", xaxt="n")
lines(satışlar$ceket, type="o", col="green", lty=2)
lines(satışlar$gömlek, type="o", col="black", lty=3)
lines(satışlar$kravat, type="o", col="red", lty=4)
text(axTicks(1), par("usr") [3]-2, adj=1, labels=c("ocak","şubat","mart","nisan","mayıs","haziraz"), xpd=T)
legend(4,152, legend=c("takım elbise", "ceket","gömlek","kravat"), col=c("blue","green","blcak","red"), lty=1:4)
```

```r
veri <- data.frame("kategori" = c('A', 'B', 'C', 'D'), "miktar"= c(25,40,27,8))
```

```r
ggplot(veri, aes(x="", y=miktar, fill=kategori)) + geom_bar(stat="identity", width=1) + coord_polar("y", start=0) + goem_text(aes(label = paste0(miktar, "%")), position = position_stack(vjust=0.5)) + labs(x=NULLi y=NULL, fill=NULL) + 
theme_calssic() + theme(axis.text= element_blank(), axis_ticks= element_blank()) + scale=fill_manual(values= c("red", "yellow","blue", "green"))
```
