```r
veri <- read.csv("C:/diabetes.csv")
```

```r
head(veri)
```

```r
str(veri)
```

```r
summary(veri)
```

```r
veri$Outcome <- as.factor(veri$Outcome)
veri$Pregnancies <- as.factor(veri$Pregnancies)
```

```r
xtabs(~Outcome + Pregnancies, data= veri)
```

```r
set.seed(1234)
ind <- sample(2, nrow(veri), replace= T, prob= c(0.8,0.2))
train <- veri[ind==1,]
test <- veri[ind==2,]
```

```r
nrow(train)
nrow(test)
```

```r
benimmodel <- glm(Outcome ~ Glucose+BloodPressure+SkinThickness+Insulin+BMI+DiabetesPedigreeFunction+Age+Pregnancies, data= train, family = 'binomial')
summary(benimmodel)
```

```r
benimmodel <- glm(Outcome ~ Glucose+BMI+Pregnancies, data= train, family = 'binomial')
summary(benimmodel)
```

```r
tahmin <- predict(benimmodel, train, type ='response')
```

```r
head(tahmin)
```

```r
tahmin1 <- ifelse(tahmin>0.5, 1, 0)
tab1 <- table(Predicted =tahmin1, Actual =train$Outcome)
```

```r
tab1
```

```r
sum(diag(tab1))/sum(tab1)
```

```r
tahmin2 <- predict(benimmodel, test, type= 'response')
test_tahmin <- ifelse(tahmin2> 0.5, 1, 0)
tab2 <- table(Predicted = test_tahmin, Actual=test$Outcome)
```

```r
tab2
```

```r
sum(diag(tab2))/sum(tab2)
```
