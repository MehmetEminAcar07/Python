```r
veri=matrix(c(7,4,2,0,9,6,3,5,4,1,7,8), nrow=6)
veri
```

```r
round(dist(veri),2)
```

```r
res.k= kmeans(veri, centers=veri[c(5,6),], algorith="Lloyd")
```

```r
res.k
```

```r
res.k$cluster #verilerin atanmış oldukları kümeler belirleniyor.
```

```r
round(res.k$centers, digits=2) #oluşan kümelerin final merkezleri belirleniyor.
```

```r
round(res.k$betweenss/res.k$totss,3)  #kümelerin başarısı için bss/tss e bölünüyor.
```
