```r
set.seed(1)
veri=sample.int(25, 30, replace=TRUE)
veri
```

```r
ortalama= mean(veri)
print(ortalama)
```

```r
medyan= median(veri)
print(medyan)
```

```r
mod_bul <- function(v) {
    uniqv <- unique(v)
    uniqv[which.max(tabulate(match(v, uniqv)))]
}
sonuc <- mod_bul(veri)
print(sonuc)
```

```r
varyans= var(veri)
varyans
standartsapma= sqrt(var(veri))
print(standartsapma)
```
